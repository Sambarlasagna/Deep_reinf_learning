```python
%%bash
mkdir -p training-envs-executables/linux
wget https://github.com/huggingface/Snowball-Target/raw/main/SnowballTarget.zip \
  -O training-envs-executables/linux/SnowballTarget.zip
unzip -o training-envs-executables/linux/SnowballTarget.zip \
  -d training-envs-executables/linux/
chmod +x training-envs-executables/linux/SnowballTarget/SnowballTarget.x86_64
```

```python
!ls training-envs-executables/linux/SnowballTarget
```

```python
%%bash
mkdir -p config/ppo
cat << 'EOF' > config/ppo/SnowballTarget.yaml
behaviors:
  SnowballTarget:
    trainer_type: ppo
    hyperparameters:
      batch_size: 1024
      buffer_size: 10240
      learning_rate: 3.0e-4
      beta: 5.0e-4
      epsilon: 0.2
      lambd: 0.95
      num_epoch: 3
    network_settings:
      normalize: false
      hidden_units: 128
      num_layers: 2
    reward_signals:
      extrinsic:
        gamma: 0.99
        strength: 1.0
    max_steps: 500000
    time_horizon: 64
    summary_freq: 10000
EOF
```

```python
!ls config/ppo
```

```python
!/usr/local/miniconda/bin/conda run -n mlagents \
python -m mlagents.trainers.learn \
config/ppo/SnowballTarget.yaml \
--env=training-envs-executables/linux/SnowballTarget/SnowballTarget.x86_64 \
--run-id=SnowballTarget1 \
--no-graphics
```
